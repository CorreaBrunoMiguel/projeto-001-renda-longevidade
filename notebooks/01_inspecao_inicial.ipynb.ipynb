## Estrutura e dimensões

## Interpretação da unidade de registro

Cada linha dos arquivos reúne a identificação de uma economia
ou agregado, o nome e o código do indicador e uma série temporal
organizada em colunas anuais de 1960 a 2025.

Nos subconjuntos sul-americanos selecionados, cada linha corresponde
a um país e ao indicador do respectivo arquivo. Os campos anuais
vazios são representados como `NaN` após a leitura pelo Pandas.

Para localizar uma observação país-ano, identifico o país pelo
seu código e seleciono a coluna do ano desejado, no conjunto
referente ao indicador.

Assim, distingo a série temporal contida na linha do valor
referente a um ano específico, que também pode estar ausente.
